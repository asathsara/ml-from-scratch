# K-Means Clustering using Scikit-Learn

This notebook demonstrates how to cluster customer spending data using Scikit-Learn's `KMeans`.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

df = pd.read_csv('customer_data.csv')
print(df.head())

# Fit KMeans
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = kmeans.fit_predict(df)

# Plot
plt.scatter(df['Annual_Income_USD'], df['Spending_Score'], c=labels, cmap='viridis')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], color='red', marker='X', s=200)
plt.xlabel('Annual Income (USD)')
plt.ylabel('Spending Score')
plt.title('Customer Clustering')
plt.show()
```
